# 4. Concatenation des DataFrames

```python
list_of_dataframes = []
for file in csv_files :
    temp_df = pd.read_csv(file)
    clean_df = []
    for col in temp_df.columns :
        new_col = col.strip() 
        clean_df.append(new_col)
    temp_df.columns = clean_df
    day = file.name[0:file.name.find("-")]
    temp_df["source_day"] = day
    list_of_dataframes.append(temp_df)
```

```python
len(list_of_dataframes)
```

```python
df_final = pd.concat(list_of_dataframes)
```

```python
df_final.shape
```

```python
df_final["Label"].value_counts()
```

```python
df_final["source_day"].value_counts()
```

# 5. Add Binary Column  

```python
df_final["is_attack"] = (df_final["Label"] != "BENIGN").astype(int)
df_final.head()
df_final["is_attack"].value_counts()
```

# 6. Clean Inf/NaN

```python
np.isinf(df_final.select_dtypes(include=np.number)).sum().sum()
```

```python
df_final.isna().sum().sum()
```

```python
np.isinf(df_final.select_dtypes(include=np.number)).sum().sum()
```

```python
df_final = df_final.replace([np.inf, -np.inf], np.nan)
```

```python
np.isinf(df_final.select_dtypes(include=np.number)).sum().sum()
```

# 7. Delete NAN Values

```python
df_final.isna().sum().sum()
```

```python
df_final = df_final.dropna()
```
